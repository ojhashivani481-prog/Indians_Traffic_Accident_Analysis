## step-3: Data cleaning

```python
# Part 3 - Data Cleaning: Check Fatalities Against Casualties

# Count records where fatalities exceed casualties
invalid_records = df[
    df["Number of Fatalities"] > df["Number of Casualties"]
]

# Display the number of records that need investigation
print("Records to investigate:", len(invalid_records))

# Display the relevant columns for checking
print(invalid_records[
    ["Number of Casualties", "Number of Fatalities"]
].head())
```

```python
# Count records with zero casualties
print(
    "Zero casualties:",
    (df["Number of Casualties"] == 0).sum()
)

# Count records with zero fatalities
print(
    "Zero fatalities:",
    (df["Number of Fatalities"] == 0).sum()
)
```

```python
#  Fill Missing Values

# Replace missing values with "Unknown"
df["Traffic Control Presence"] = df["Traffic Control Presence"].fillna("Unknown")

# Replace missing values with "Unknown"
df["Driver License Status"] = df["Driver License Status"].fillna("Unknown")

# Check missing values after cleaning
print(df[["Traffic Control Presence", "Driver License Status"]].isnull().sum())
```

```python
#  Check Unknown City Names

# Count city records marked as Unknown
print("Unknown city records:", (df["City Name"] == "Unknown").sum())
```

```python
#  Check Remaining Missing Values

# Count missing values in every column
print(df.isnull().sum())
```

## Step-4:  Exploratory Data Analysis (EDA)

EDA kya hota hai?

EDA ka full form Exploratory Data Analysis hai. Ismein hum dataset ko analyze karke patterns aur useful information nikaalte hain.

```python
# Load the Cleaned Dataset

# Load the cleaned CSV file into a DataFrame
df = pd.read_csv("../data/cleaned_accident_data.csv")

# Display the number of rows and columns
print("Dataset shape:", df.shape)

# Display the first five records
print(df.head())
```

```python
# Analyze Accident Severity

# Count records in each accident severity category
severity_counts = df["Accident Severity"].value_counts()

# Display the results
print(severity_counts)
```

```python
# Part 4 - Create a High-Quality Accident Severity Chart

import matplotlib.pyplot as plt



# Create the bar chart
plt.bar(
    severity_counts.index,
    severity_counts.values,
    color="pink",
    edgecolor="black",
    width=0.6
)

# Make the title and labels easier to read
plt.title("Accident Severity Analysis", fontsize=16, fontweight="bold")
plt.xlabel("Accident Severity", fontsize=13)
plt.ylabel("Number of Records", fontsize=13)

# Increase the size of category labels
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)

# Add a light grid
plt.grid(axis="y", linestyle="--", alpha=0.3)

# Adjust spacing
plt.tight_layout()

# Save the chart with high resolution
plt.savefig(
    "../charts/accident_severity.png",
    dpi=300,
    bbox_inches="tight"
)

# Display and close the chart
plt.show()
plt.close()
```

```python
#  Analyze Weather Conditions

# Count records for each weather condition
weather_counts = df["Weather Conditions"].value_counts()

# Display the results
print(weather_counts)
```

```python

#  Weather Condition Chart


# Create the pink bar chart
plt.bar(
    weather_counts.index,
    weather_counts.values,
    color="lime",
    edgecolor="black"
)

# Add chart title and axis labels
plt.title("Accident Records by Weather Condition")
plt.xlabel("Weather Condition")
plt.ylabel("Number of Records")

# Rotate labels to make them easier to read
plt.xticks(rotation=45)

# Add a light horizontal grid
plt.grid(axis="y", linestyle="--", alpha=0.3)

# Adjust spacing around the chart
plt.tight_layout()

# Save the chart inside the charts folder
plt.savefig(
    "../charts/weather_condition.png",
    dpi=300,
    bbox_inches="tight"
)

# Display the chart
plt.show()

# Close the figure after displaying it
plt.close()
```

```python

#  Average Casualties by Accident Severity

# Calculate average casualties for each severity
severity_casualties = df.groupby("Accident Severity")["Number of Casualties"].mean()

# Display the results
print(severity_casualties)
```

```python

#  Seaborn Casualties Chart

import seaborn as sns
import matplotlib.pyplot as plt

# Create a bar chart using Seaborn
sns.barplot(
    x=severity_casualties.index,
    y=severity_casualties.values,
    color="lawngreen"
)

# Add chart title and labels
plt.title("Average Casualties by Accident Severity")
plt.xlabel("Accident Severity")
plt.ylabel("Average Number of Casualties")

# Add a light grid
plt.grid(axis="y", linestyle="--", alpha=0.3)

# Adjust chart spacing
plt.tight_layout()

# Save the chart
plt.savefig(
    "../charts/severity_casualties.png",
    dpi=300,
    bbox_inches="tight"
)

# Display the chart
plt.show()

# Close the figure
plt.close()
```

```python

# Day-wise Accident Chart


# Create the bar chart
sns.barplot(
    x=day_counts.index,
    y=day_counts.values,
    color="hotpink"
)

# Add chart title and labels
plt.title("Accident Records by Day of Week")
plt.xlabel("Day of Week")
plt.ylabel("Number of Records")

# Rotate day names for readability
plt.xticks(rotation=30)

# Add a light horizontal grid
plt.grid(axis="y", linestyle="--", alpha=0.3)

# Adjust chart spacing
plt.tight_layout()

# Save the chart
plt.savefig(
    "../charts/day_wise_accidents.png",
    dpi=300,
    bbox_inches="tight"
)

# Display the chart
plt.show()

# Close the figure
plt.close()
```
